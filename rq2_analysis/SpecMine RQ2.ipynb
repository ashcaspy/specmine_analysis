```python
import pandas as pd

cols = [
    "file_url_sha16",
    "repo_name",
    "file_path",
    "seq_from_oldest",
    "message",
    "author_name",
    "author_login",
    "author_type",
    "committer_name",
    "committer_login",
    "committer_type"
]

df = commits.select_columns(cols).to_pandas()

print("Commit rows:", len(df))
print("Unique specs:", df["file_url_sha16"].nunique())

print("\nAUTHOR TYPES")
print(df["author_type"].value_counts(dropna=False))

print("\nCOMMITTER TYPES")
print(df["committer_type"].value_counts(dropna=False))
```

```python
bot_rows = df[
    (df["author_type"].fillna("").str.lower() == "bot") |
    (df["committer_type"].fillna("").str.lower() == "bot")
]

print("Bot-related commit rows:", len(bot_rows))
print("Specs with bot-related commits:", bot_rows["file_url_sha16"].nunique())

print("\nTOP BOT AUTHORS")
print(
    bot_rows["author_login"]
    .value_counts(dropna=False)
    .head(50)
)

print("\nTOP BOT COMMITTERS")
print(
    bot_rows["committer_login"]
    .value_counts(dropna=False)
    .head(50)
)
```

```python
ai_pattern = (
    r"claude|anthropic|copilot|chatgpt|openai|gemini|"
    r"cursor|codex|jules|devin|"
    r"ai[- ]generated|generated by ai|generated with|"
    r"co-authored-by:.*(?:claude|anthropic|copilot|chatgpt|openai|gemini|cursor|codex|jules|devin)"
)

ai_mentions = df[
    df["message"].fillna("").str.contains(
        ai_pattern,
        case=False,
        regex=True
    )
]

print("Commit rows with possible AI evidence:", len(ai_mentions))
print("Unique specs with possible AI evidence:",
      ai_mentions["file_url_sha16"].nunique())

display(
    ai_mentions[
        [
            "file_url_sha16",
            "repo_name",
            "seq_from_oldest",
            "author_login",
            "author_type",
            "committer_login",
            "message"
        ]
    ].head(100)
)
```

```python
ai_account_pattern = (
    r"copilot|claude|codex|jules|devin|cursor|"
    r"lovable|kilo-code|opencode-agent|"
    r"fullsend-ai|aiagent|agnt-platform"
)

ai_account_rows = df[
    df["author_login"].fillna("").str.contains(
        ai_account_pattern, case=False, regex=True
    )
    |
    df["committer_login"].fillna("").str.contains(
        ai_account_pattern, case=False, regex=True
    )
]

print("Possible AI-agent commit rows:", len(ai_account_rows))
print(
    "Possible AI-agent specs:",
    ai_account_rows["file_url_sha16"].nunique()
)

print("\nAUTHOR ACCOUNTS")
print(
    ai_account_rows["author_login"]
    .value_counts(dropna=False)
    .head(50)
)

print("\nCOMMITTER ACCOUNTS")
print(
    ai_account_rows["committer_login"]
    .value_counts(dropna=False)
    .head(50)
)
```

```python
print("AUTHOR ACCOUNTS")
print(
    ai_account_rows["author_login"]
    .value_counts(dropna=False)
    .head(30)
    .to_string()
)

print("\nCOMMITTER ACCOUNTS")
print(
    ai_account_rows["committer_login"]
    .value_counts(dropna=False)
    .head(30)
    .to_string()
)
```

```python
HIGH_CONF_AI = {
    "claude",
    "copilot",
    "google-labs-jules[bot]",
    "cursoragent",
    "codex-scheduled-tasks[bot]",
    "d-sorgcodexagent[bot]",
    "codex",
    "d-sorgclaudeagent[bot]",
    "lovable-dev[bot]",
    "fullsend-ai-coder[bot]",
    "claude[bot]",
    "agnt-platform[bot]",
    "devin-ai-integration[bot]",
    "cursor[bot]",
    "jessica-claude[bot]",
    "aiagentxinlocal[bot]",
    "kilo-code-bot[bot]",
    "opencode-agent[bot]",
}

HIGH_CONF_AI = {x.lower() for x in HIGH_CONF_AI}

author = df["author_login"].fillna("").str.lower()
committer = df["committer_login"].fillna("").str.lower()

df["ai_account_evidence"] = (
    author.isin(HIGH_CONF_AI) |
    committer.isin(HIGH_CONF_AI)
)

print("High-confidence AI-related commits:",
      df["ai_account_evidence"].sum())

print("Specs with high-confidence AI evidence:",
      df.loc[df["ai_account_evidence"],
             "file_url_sha16"].nunique())
```

```python
author_login_norm = df["author_login"].fillna("").str.lower()

df["human_author_evidence"] = (
    (df["author_type"].fillna("").str.lower() == "user")
    & (author_login_norm != "")
    & (~author_login_norm.isin(HIGH_CONF_AI))
)

spec_authorship = (
    df.groupby("file_url_sha16")
      .agg(
          has_ai_evidence=("ai_account_evidence", "max"),
          has_human_evidence=("human_author_evidence", "max"),
          n_commits=("file_url_sha16", "size"),
          first_seq=("seq_from_oldest", "min"),
          last_seq=("seq_from_oldest", "max")
      )
      .reset_index()
)

def classify_authorship(row):
    if row["has_ai_evidence"] and row["has_human_evidence"]:
        return "Collaborative"
    elif row["has_ai_evidence"] and not row["has_human_evidence"]:
        return "AI-written"
    elif not row["has_ai_evidence"] and row["has_human_evidence"]:
        return "Human-written"
    else:
        return "Unknown"

spec_authorship["authorship_group"] = spec_authorship.apply(
    classify_authorship,
    axis=1
)

print("TOTAL SPECS WITH COMMIT HISTORY:",
      len(spec_authorship))

print("\nAUTHORSHIP GROUP COUNTS")
print(
    spec_authorship["authorship_group"]
    .value_counts(dropna=False)
)

print("\nAUTHORSHIP GROUP PERCENTAGES")
print(
    (
        spec_authorship["authorship_group"]
        .value_counts(normalize=True)
        * 100
    ).round(2)
)
```

```python
import pandas as pd

model = pd.read_csv("modeling_dataset.csv")

print(model.shape)
print(model.columns.tolist())
```

```python
scores = model[
    [
        "file_url_sha16",
        "target_spec_completeness",
        "target_spec_quality"
    ]
].copy()

scores = scores.drop_duplicates(subset="file_url_sha16")

rq2 = spec_authorship.merge(
    scores,
    on="file_url_sha16",
    how="inner"
)

print("Matched RQ2 specifications:", len(rq2))

print("\nGROUP COUNTS AFTER JOIN")
print(rq2["authorship_group"].value_counts())

print("\nMISSING SCORES")
print(
    rq2[
        ["target_spec_completeness",
         "target_spec_quality"]
    ].isna().sum()
)
```

```python
analysis_rq2 = rq2[
    rq2["authorship_group"].isin(
        ["Human-written", "AI-written", "Collaborative"]
    )
].copy()

summary = (
    analysis_rq2
    .groupby("authorship_group")
    .agg(
        N=("file_url_sha16", "count"),

        completeness_mean=(
            "target_spec_completeness", "mean"
        ),
        completeness_median=(
            "target_spec_completeness", "median"
        ),
        completeness_sd=(
            "target_spec_completeness", "std"
        ),

        quality_mean=(
            "target_spec_quality", "mean"
        ),
        quality_median=(
            "target_spec_quality", "median"
        ),
        quality_sd=(
            "target_spec_quality", "std"
        )
    )
    .round(3)
)

display(summary)
```

```python
from scipy.stats import kruskal, mannwhitneyu
from statsmodels.stats.multitest import multipletests
import pandas as pd

groups = ["Human-written", "AI-written", "Collaborative"]

results_kw = {}

for outcome in [
    "target_spec_completeness",
    "target_spec_quality"
]:
    samples = [
        analysis_rq2.loc[
            analysis_rq2["authorship_group"] == g,
            outcome
        ].dropna()
        for g in groups
    ]

    H, p = kruskal(*samples)


    N = sum(len(x) for x in samples)
    k = len(samples)
    epsilon_sq = (H - k + 1) / (N - k)

    results_kw[outcome] = {
        "H": H,
        "p": p,
        "epsilon_squared": epsilon_sq
    }

print("KRUSKAL-WALLIS RESULTS")

for outcome, r in results_kw.items():
    print("\n", outcome)
    print("H =", round(r["H"], 3))
    print("p =", r["p"])
    print("epsilon squared =", round(r["epsilon_squared"], 6))



pairs = [
    ("Human-written", "AI-written"),
    ("Human-written", "Collaborative"),
    ("AI-written", "Collaborative")
]

for outcome in [
    "target_spec_completeness",
    "target_spec_quality"
]:

    pair_results = []

    for g1, g2 in pairs:

        x = analysis_rq2.loc[
            analysis_rq2["authorship_group"] == g1,
            outcome
        ].dropna()

        y = analysis_rq2.loc[
            analysis_rq2["authorship_group"] == g2,
            outcome
        ].dropna()

        U, p = mannwhitneyu(
            x,
            y,
            alternative="two-sided"
        )

        pair_results.append({
            "comparison": f"{g1} vs {g2}",
            "U": U,
            "raw_p": p
        })

    raw_ps = [r["raw_p"] for r in pair_results]

    reject, corrected_ps, _, _ = multipletests(
        raw_ps,
        method="holm"
    )

    for i, r in enumerate(pair_results):
        r["holm_p"] = corrected_ps[i]
        r["significant"] = reject[i]

    print("\nPAIRWISE:", outcome)
    display(pd.DataFrame(pair_results))
```
